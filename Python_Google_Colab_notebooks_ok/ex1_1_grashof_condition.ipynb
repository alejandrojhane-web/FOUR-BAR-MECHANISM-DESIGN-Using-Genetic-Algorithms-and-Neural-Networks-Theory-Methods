## Statement
There are four links with lengths:
- $L_1 = 100$ mm (frame)
- $L_2 = 40$ mm (crank)
- $L_3 = 80$ mm (coupler)
- $L_4 = 60$ mm (rocker)

Determine if the mechanism satisfies the **Grashof condition** and classify the resulting type of mechanism.

## Theoretical foundation
The Grashof condition states that at least one of the links can rotate 360° if:

$$S + L \leq P + Q$$

where $S$ is the shortest link, $L$ the longest, and $P$, $Q$ the two in between.

| Shortest link position | Mechanism type |
|---|---|
| Frame ($L_1$) or coupler ($L_3$) | Crank-rocker |
| Crank ($L_2$) | Double-crank (drag-link) |
| Rocker ($L_4$) | Double-rocker |

```python
from numpy import *

# Link lengths (mm)
L1, L2, L3, L4 = 100.0, 40.0, 80.0, 60.0
longitudes = array([L1, L2, L3, L4])

S = longitudes.min()          # shortest link
L = longitudes.max()          # longest link
idx_otros = argsort(longitudes)[1:3]
P, Q = longitudes[idx_otros]  # the two intermediate ones

print(f"S + L = {S + L:.1f} mm")
print(f"P + Q = {P + Q:.1f} mm")

if S + L <= P + Q:
    print("Grashof Condition: SATISFIED")
    if longitudes[0] == S or longitudes[3] == S:
        print("Type: Crank-rocker")
    elif longitudes[1] == S:
        print("Type: Double-crank")
    else:
        print("Type: Double-rocker")
else:
    print("Grashof Condition: NOT satisfied → rocker-rocker mechanism")
```
